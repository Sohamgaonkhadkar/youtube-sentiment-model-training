```python
# ========================= IMPORTS =========================
import mlflow
import mlflow.sklearn
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    classification_report,
    confusion_matrix,
    f1_score
)
```

```python
df = pd.read_csv('reddit_preprocessing.csv').dropna(subset=['clean_comment'])
df.shape
```

```python
# =========================================================
# RESULTS STORAGE
# =========================================================
results = []

# =========================================================
# FUNCTION
# =========================================================
def run_experiment_tfidf_max_features(max_features):

    # -----------------------------------------------------
    # NGRAM
    # -----------------------------------------------------
    ngram_range = (1, 3)

    # -----------------------------------------------------
    # TF-IDF VECTORIZER
    # -----------------------------------------------------
    vectorizer = TfidfVectorizer(
        ngram_range=ngram_range,
        max_features=max_features
    )
    
    X_train, X_test, y_train, y_test = train_test_split(
        df['clean_comment'],
        df['category'],
        test_size=0.2,
        random_state=42,
        stratify=df['category']
    )

    # -----------------------------------------------------
    # VECTORIZE
    # -----------------------------------------------------
    X_train_vec = vectorizer.fit_transform(
        X_train
    )

    X_test_vec = vectorizer.transform(
        X_test
    )

    # -----------------------------------------------------
    # START RUN
    # -----------------------------------------------------
    with mlflow.start_run():

        # -------------------------------------------------
        # RUN NAME
        # -------------------------------------------------
        mlflow.set_tag(
            "mlflow.runName",
            f"TFIDF_Trigram_MaxFeatures_{max_features}"
        )

        mlflow.set_tag(
            "experiment_type",
            "feature_engineering"
        )

        mlflow.set_tag(
            "model_type",
            "RandomForest"
        )

        mlflow.set_tag(
            "description",
            f"TFIDF trigram RF with max_features={max_features}"
        )

        # -------------------------------------------------
        # PARAMETERS
        # -------------------------------------------------
        mlflow.log_param(
            "vectorizer_type",
            "TF-IDF"
        )

        mlflow.log_param(
            "ngram_range",
            str(ngram_range)
        )

        mlflow.log_param(
            "max_features",
            max_features
        )

        # -------------------------------------------------
        # MODEL PARAMETERS
        # -------------------------------------------------
        n_estimators = 200
        max_depth = 15

        mlflow.log_param(
            "n_estimators",
            n_estimators
        )

        mlflow.log_param(
            "max_depth",
            max_depth
        )

        # -------------------------------------------------
        # MODEL
        # -------------------------------------------------
        model = RandomForestClassifier(
            n_estimators=n_estimators,
            max_depth=max_depth,
            random_state=42,
            n_jobs=-1
        )

        # -------------------------------------------------
        # TRAIN
        # -------------------------------------------------
        model.fit(
            X_train_vec,
            y_train
        )

        # -------------------------------------------------
        # PREDICT
        # -------------------------------------------------
        y_pred = model.predict(
            X_test_vec
        )

        # -------------------------------------------------
        # METRICS
        # -------------------------------------------------
        accuracy = accuracy_score(
            y_test,
            y_pred
        )

        weighted_f1 = f1_score(
            y_test,
            y_pred,
            average='weighted'
        )

        macro_f1 = f1_score(
            y_test,
            y_pred,
            average='macro'
        )

        vocab_size = len(
            vectorizer.vocabulary_
        )

        # -------------------------------------------------
        # LOG METRICS
        # -------------------------------------------------
        mlflow.log_metric(
            "accuracy",
            accuracy
        )

        mlflow.log_metric(
            "weighted_f1",
            weighted_f1
        )

        mlflow.log_metric(
            "macro_f1",
            macro_f1
        )

        mlflow.log_metric(
            "vocab_size",
            vocab_size
        )

        # -------------------------------------------------
        # CLASSIFICATION REPORT
        # -------------------------------------------------
        classification_rep = classification_report(
            y_test,
            y_pred,
            output_dict=True
        )

        for label, metrics in classification_rep.items():

            if isinstance(metrics, dict):

                for metric, value in metrics.items():

                    mlflow.log_metric(
                        f"{label}_{metric}",
                        value
                    )

        # =================================================
        # CONFUSION MATRIX
        # =================================================
        conf_matrix = confusion_matrix(
            y_test,
            y_pred
        )

        fig, ax = plt.subplots(
            figsize=(8, 6)
        )

        sns.heatmap(
            conf_matrix,
            annot=True,
            fmt="d",
            cmap="Blues",
            ax=ax
        )

        ax.set_xlabel("Predicted")
        ax.set_ylabel("Actual")

        ax.set_title(
            f"TFIDF Trigram\nmax_features={max_features}"
        )

        mlflow.log_figure(
            fig,
            f"conf_matrix_{max_features}.png"
        )

        plt.close(fig)

        # =================================================
        # FEATURE IMPORTANCE
        # =================================================
        feature_importance = (
            model.feature_importances_
        )

        top_n = 20

        indices = (
            feature_importance.argsort()[-top_n:]
        )

        feature_names = (
            vectorizer.get_feature_names_out()
        )

        top_features = feature_names[indices]

        top_importance = (
            feature_importance[indices]
        )

        fig, ax = plt.subplots(
            figsize=(10, 6)
        )

        ax.barh(
            top_features,
            top_importance
        )

        ax.set_title(
            f"Top Features\nmax_features={max_features}"
        )

        ax.set_xlabel(
            "Importance"
        )

        mlflow.log_figure(
            fig,
            f"feature_importance_{max_features}.png"
        )

        plt.close(fig)

        # =================================================
        # SAVE MODEL
        # =================================================
        mlflow.sklearn.log_model(
            sk_model=model,
            name=f"rf_model_tfidf_{max_features}"
        )

        # =================================================
        # SAVE VECTORIZER
        # =================================================
        vectorizer_filename = (
            f"vectorizer_tfidf_{max_features}.pkl"
        )

        joblib.dump(
            vectorizer,
            vectorizer_filename
        )

        mlflow.log_artifact(
            vectorizer_filename
        )

        # =================================================
        # STORE RESULTS
        # =================================================
        results.append({

            "Max_Features": max_features,

            "Accuracy": accuracy,

            "Weighted_F1": weighted_f1,

            "Macro_F1": macro_f1,

            "Vocab_Size": vocab_size
        })

        print("=" * 60)
        print(f"MAX FEATURES: {max_features}")
        print(f"Accuracy: {accuracy:.4f}")
        print(f"Weighted F1: {weighted_f1:.4f}")
        print(f"Macro F1: {macro_f1:.4f}")
        print("=" * 60)


# =========================================================
# RUN EXPERIMENTS
# =========================================================
max_features_values = [
    1000,
    2000,
    3000,
    4000,
    5000,
    6000,
    7000,
    8000,
    9000,
    10000
]

for max_features in max_features_values:

    run_experiment_tfidf_max_features(
        max_features
    )

# =========================================================
# RESULTS DATAFRAME
# =========================================================
results_df = pd.DataFrame(results)

print(results_df)

# =========================================================
# COMPARISON PLOT
# =========================================================
fig, ax = plt.subplots(
    figsize=(12, 6)
)

sns.lineplot(
    data=results_df,
    x="Max_Features",
    y="Accuracy",
    marker="o",
    linewidth=3,
    ax=ax
)

ax.set_title(
    "TFIDF Max Features vs Accuracy",
    fontsize=16
)

ax.set_xlabel(
    "Max Features",
    fontsize=12
)

ax.set_ylabel(
    "Accuracy",
    fontsize=12
)

ax.grid(True)

# =========================================================
# SAVE COMPARISON RUN
# =========================================================
with mlflow.start_run(
    run_name="TFIDF_MaxFeatures_Comparison"
):

    mlflow.log_figure(
        fig,
        "overall_max_features_comparison.png"
    )

plt.show()

plt.close(fig)
```

```python
# =========================================================
# IMPORTS
# =========================================================
import pandas as pd
import numpy as np
import mlflow
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_validate

# =========================================================
# RESULTS STORAGE
# =========================================================
results = []

# =========================================================
# CROSS VALIDATION
# =========================================================
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# =========================================================
# FUNCTION
# =========================================================
def run_experiment_tfidf_max_features(max_features):

    print("=" * 60)
    print(f"RUNNING FOR MAX_FEATURES = {max_features}")
    print("=" * 60)

    # -----------------------------------------------------
    # PIPELINE
    # -----------------------------------------------------
    pipeline = Pipeline([

        (
            "tfidf",
            TfidfVectorizer(
                ngram_range=(1, 3),
                max_features=max_features
            )
        ),

        (
            "model",
            LinearSVC(
                class_weight='balanced',
                random_state=42
            )
        )
    ])

    # -----------------------------------------------------
    # CROSS VALIDATION
    # -----------------------------------------------------
    scores = cross_validate(

        pipeline,

        df['clean_comment'],
        df['category'],

        cv=cv,

        scoring={

            'accuracy': 'accuracy',

            'macro_f1': 'f1_macro',

            'weighted_f1': 'f1_weighted'
        },

        n_jobs=-1,
        return_train_score=False
    )

    # -----------------------------------------------------
    # MEAN SCORES
    # -----------------------------------------------------
    accuracy = np.mean(
        scores['test_accuracy']
    )

    macro_f1 = np.mean(
        scores['test_macro_f1']
    )

    weighted_f1 = np.mean(
        scores['test_weighted_f1']
    )

    # -----------------------------------------------------
    # STORE RESULTS
    # -----------------------------------------------------
    results.append({

        "Max_Features": max_features,

        "Accuracy": accuracy,

        "Macro_F1": macro_f1,

        "Weighted_F1": weighted_f1
    })

    # -----------------------------------------------------
    # MLFLOW
    # -----------------------------------------------------
    with mlflow.start_run():

        mlflow.set_tag(
            "experiment_type",
            "TFIDF_MaxFeatures_CV"
        )

        mlflow.set_tag(
            "model_type",
            "LinearSVM"
        )

        mlflow.log_param(
            "max_features",
            max_features
        )

        mlflow.log_param(
            "ngram_range",
            "(1,3)"
        )

        mlflow.log_param(
            "cv_folds",
            5
        )

        mlflow.log_metric(
            "accuracy",
            accuracy
        )

        mlflow.log_metric(
            "macro_f1",
            macro_f1
        )

        mlflow.log_metric(
            "weighted_f1",
            weighted_f1
        )

    # -----------------------------------------------------
    # PRINT
    # -----------------------------------------------------
    print(f"Accuracy    : {accuracy:.4f}")
    print(f"Macro F1    : {macro_f1:.4f}")
    print(f"Weighted F1 : {weighted_f1:.4f}")

# =========================================================
# VALUES
# =========================================================
max_features_values = [

    1000,
    2000,
    3000,
    5000,
    7000,
    10000
]

# =========================================================
# RUN
# =========================================================
for max_features in max_features_values:

    run_experiment_tfidf_max_features(
        max_features
    )

# =========================================================
# RESULTS DATAFRAME
# =========================================================
results_df = pd.DataFrame(results)

print("\nFINAL RESULTS")
print(results_df)

# =========================================================
# SORT BY MACRO F1
# =========================================================
results_df = results_df.sort_values(
    by="Macro_F1",
    ascending=False
)

print("\nSORTED BY MACRO F1")
print(results_df)

# =========================================================
# PLOT - MACRO F1
# =========================================================
fig, ax = plt.subplots(
    figsize=(12, 6)
)

sns.lineplot(

    data=results_df,

    x="Max_Features",
    y="Macro_F1",

    marker="o",
    linewidth=3,
    ax=ax
)

ax.set_title(
    "TFIDF Max Features vs Macro F1",
    fontsize=16
)

ax.set_xlabel(
    "Max Features",
    fontsize=12
)

ax.set_ylabel(
    "Macro F1",
    fontsize=12
)

ax.grid(True)

plt.show()

# =========================================================
# BEST RESULT
# =========================================================
best_row = results_df.iloc[0]

print("\nBEST CONFIGURATION")
print(best_row)
```

```python
# =========================================================
# IMPORTS
# =========================================================
import pandas as pd
import numpy as np
import mlflow
import matplotlib.pyplot as plt
import seaborn as sns

from lightgbm import LGBMClassifier

from sklearn.pipeline import Pipeline
from sklearn.model_selection import (
    StratifiedKFold,
    cross_validate
)

from sklearn.feature_extraction.text import (
    TfidfVectorizer
)

# =========================================================
# RESULTS STORAGE
# =========================================================
results = []

# =========================================================
# CROSS VALIDATION
# =========================================================
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# =========================================================
# FUNCTION
# =========================================================
def run_lgbm_experiment(max_features):

    print("=" * 60)
    print(f"RUNNING LIGHTGBM : {max_features}")
    print("=" * 60)

    # -----------------------------------------------------
    # PIPELINE
    # -----------------------------------------------------
    pipeline = Pipeline([

        (
            "tfidf",

            TfidfVectorizer(

                ngram_range=(1,3),

                max_features=max_features,

                min_df=2,

                max_df=0.95
            )
        ),

        (
            "model",

            LGBMClassifier(

                n_estimators=300,

                learning_rate=0.05,

                max_depth=10,

                num_leaves=31,

                subsample=0.8,

                colsample_bytree=0.8,

                class_weight='balanced',

                random_state=42,

                n_jobs=-1
            )
        )
    ])

    # -----------------------------------------------------
    # CROSS VALIDATION
    # -----------------------------------------------------
    scores = cross_validate(

        pipeline,

        df['clean_comment'],
        df['category'],

        cv=cv,

        scoring={

            'accuracy': 'accuracy',

            'macro_f1': 'f1_macro',

            'weighted_f1': 'f1_weighted'
        },

        n_jobs=-1
    )

    # -----------------------------------------------------
    # MEAN SCORES
    # -----------------------------------------------------
    accuracy = np.mean(
        scores['test_accuracy']
    )

    macro_f1 = np.mean(
        scores['test_macro_f1']
    )

    weighted_f1 = np.mean(
        scores['test_weighted_f1']
    )

    # -----------------------------------------------------
    # STORE RESULTS
    # -----------------------------------------------------
    results.append({

        "Max_Features": max_features,

        "Accuracy": accuracy,

        "Macro_F1": macro_f1,

        "Weighted_F1": weighted_f1
    })

    # -----------------------------------------------------
    # LOGGING
    # -----------------------------------------------------
    with mlflow.start_run():

        mlflow.set_tag(
            "model",
            "LightGBM"
        )

        mlflow.log_param(
            "max_features",
            max_features
        )

        mlflow.log_metric(
            "accuracy",
            accuracy
        )

        mlflow.log_metric(
            "macro_f1",
            macro_f1
        )

        mlflow.log_metric(
            "weighted_f1",
            weighted_f1
        )

    # -----------------------------------------------------
    # PRINT
    # -----------------------------------------------------
    print(f"Accuracy    : {accuracy:.4f}")
    print(f"Macro F1    : {macro_f1:.4f}")
    print(f"Weighted F1 : {weighted_f1:.4f}")

# =========================================================
# VALUES
# =========================================================
max_features_values = [

    3000,
    5000,
    7000
]

# =========================================================
# RUN
# =========================================================
for max_features in max_features_values:

    run_lgbm_experiment(
        max_features
    )

# =========================================================
# RESULTS
# =========================================================
results_df = pd.DataFrame(results)

print("\nFINAL RESULTS")
print(results_df)

# =========================================================
# PLOT
# =========================================================
fig, ax = plt.subplots(
    figsize=(10,6)
)

sns.lineplot(

    data=results_df,

    x="Max_Features",

    y="Macro_F1",

    marker="o",

    linewidth=3,

    ax=ax
)

ax.set_title(
    "LightGBM TFIDF Max Features vs Macro F1"
)

ax.grid(True)

plt.show()

# =========================================================
# BEST RESULT
# =========================================================
best_row = results_df.sort_values(

    by="Macro_F1",

    ascending=False

).iloc[0]

print("\nBEST CONFIG:")
print(best_row)
```

```python
# =========================================================
# IMPORTS
# =========================================================
import pandas as pd
import numpy as np
import mlflow
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.svm import LinearSVC
from sklearn.pipeline import Pipeline
from sklearn.model_selection import StratifiedKFold, cross_validate, train_test_split



# =========================================================
# *** FIX 2: CREATE A TRUE HOLD-OUT TEST SET ***
# =========================================================
# We reserve 20% of the data completely untouched for final evaluation.
df_train, df_test = train_test_split(
    df, 
    test_size=0.2, 
    random_state=42, 
    stratify=df['category']
)

# =========================================================
# RESULTS STORAGE
# =========================================================
cv_results = []

# =========================================================
# CROSS VALIDATION SETUP (ONLY ON TRAIN DATA)
# =========================================================
cv = StratifiedKFold(
    n_splits=5,
    shuffle=True,
    random_state=42
)

# =========================================================
# EXPERIMENT FUNCTION
# =========================================================
def run_experiment_tfidf_max_features(max_features):
    print(f"Running CV for max_features = {max_features}...")

    pipeline = Pipeline([
        (
            "tfidf",
            TfidfVectorizer(
                ngram_range=(1, 3),
                max_features=max_features
            )
        ),
        (
            "model",
            LinearSVC(
                class_weight='balanced',
                random_state=42,
                dual="auto" # Added to suppress modern sklearn warnings
            )
        )
    ])

    # Run CV only on the training set
    scores = cross_validate(
        pipeline,
        df_train['clean_comment'],
        df_train['category'],
        cv=cv,
        scoring={
            'accuracy': 'accuracy',
            'macro_f1': 'f1_macro',
            'weighted_f1': 'f1_weighted'
        },
        n_jobs=-1,
        return_train_score=False
    )

    accuracy = np.mean(scores['test_accuracy'])
    macro_f1 = np.mean(scores['test_macro_f1'])
    weighted_f1 = np.mean(scores['test_weighted_f1'])

    cv_results.append({
        "Max_Features": max_features,
        "CV_Accuracy": accuracy,
        "CV_Macro_F1": macro_f1,
        "CV_Weighted_F1": weighted_f1
    })

# =========================================================
# HYPERPARAMETER SWEEP
# =========================================================
print("\n" + "=" * 60)
print("STARTING CROSS-VALIDATION SWEEP")
print("=" * 60)

max_features_values = [1000, 2000, 3000, 5000, 7000, 10000]

for max_features in max_features_values:
    run_experiment_tfidf_max_features(max_features)

# =========================================================
# SELECT BEST CONFIGURATION
# =========================================================
cv_results_df = pd.DataFrame(cv_results)
cv_results_df = cv_results_df.sort_values(by="CV_Macro_F1", ascending=False)

best_config = cv_results_df.iloc[0]
best_max_features = int(best_config["Max_Features"])

print("\n" + "=" * 60)
print("BEST CROSS-VALIDATION CONFIGURATION:")
print(best_config)
print("=" * 60)

# =========================================================
# FINAL MODEL TRAINING & TRUE TEST SET EVALUATION
# =========================================================
print(f"\nTraining final model on entire train set using max_features={best_max_features}...")

final_pipeline = Pipeline([
    ("tfidf", TfidfVectorizer(ngram_range=(1, 3), max_features=best_max_features)),
    ("model", LinearSVC(class_weight='balanced', random_state=42, dual="auto"))
])

# Train on all 80% of the training data
final_pipeline.fit(df_train['clean_comment'], df_train['category'])

# Predict on the locked 20% test data
print("Evaluating on unseen test set...")
y_pred = final_pipeline.predict(df_test['clean_comment'])
y_test = df_test['category']

# Final Metrics
test_accuracy = accuracy_score(y_test, y_pred)
test_macro_f1 = f1_score(y_test, y_pred, average='macro')
test_weighted_f1 = f1_score(y_test, y_pred, average='weighted')

print("\nFINAL TEST METRICS:")
print(f"Test Accuracy    : {test_accuracy:.4f}")
print(f"Test Macro F1    : {test_macro_f1:.4f}")
print(f"Test Weighted F1 : {test_weighted_f1:.4f}")

print("\nClassification Report:\n")
print(classification_report(y_test, y_pred))

# =========================================================
# CONFUSION MATRIX PLOT
# =========================================================
conf_matrix = confusion_matrix(y_test, y_pred)
fig, ax = plt.subplots(figsize=(8, 6))
sns.heatmap(
    conf_matrix, annot=True, fmt='d', cmap='Blues',
    xticklabels=[0, 1, 2], yticklabels=[0, 1, 2], ax=ax
)
ax.set_xlabel("Predicted")
ax.set_ylabel("Actual")
ax.set_title(f"Final Test Confusion Matrix (Max Features: {best_max_features})")
plt.show()

# =========================================================
# MLFLOW LOGGING (Logging the Final Result)
# =========================================================
with mlflow.start_run(run_name="TFIDF_LinearSVC_Final"):
    
    mlflow.set_tag("experiment_type", "TFIDF_Final_Evaluation")
    mlflow.set_tag("model_type", "LinearSVM")
    
    # Log parameters
    mlflow.log_param("best_max_features", best_max_features)
    mlflow.log_param("ngram_range", "(1,3)")
    
    # Log CV phase results
    mlflow.log_metric("cv_best_macro_f1", best_config["CV_Macro_F1"])
    
    # Log Final Test Phase metrics
    mlflow.log_metric("test_accuracy", test_accuracy)
    mlflow.log_metric("test_macro_f1", test_macro_f1)
    mlflow.log_metric("test_weighted_f1", test_weighted_f1)
    
    # Log artifacts
    mlflow.log_figure(fig, "final_test_confusion_matrix.png")

print("\nRun complete! Results logged to MLflow.")
```
